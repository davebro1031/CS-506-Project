```python
# we have N points, want K clusters, and must have at least M points in each cluster

def subarrays(start, N, M, K):
    if K == 1:
        return [[start]]

    if M*K > N-start: #breaking criterion -- you've requested an infeasible input
        return []
    
    arrays = []
    for i in range(start+M, N-(K-1)*M+1):
        curr = subarrays(i, N, M, K-1)
        for arr in curr:
            arr.append(start)
        arrays.extend(curr)

    return arrays
```

```python
result = subarrays(0,20,5,3)
print(len(result))

# this number should be nCr(N - MK + K - 1, K - 1)
```
